# Creating a Dataset Manually

```python
# Creating a Dataset Manually
import pandas as pd

# Define columns of the dataset
names = ['Alice', 'Bob', 'Charlie', ]
ages = [25, 30, 35 ]
cities = ['New York', 'Los Angeles', 'Chicago' ]
salaries = [60000, 80000, 75000]

# Create a dictionary to store the data
data = {
    'Name': names,
    'Age': ages,
    'City': cities,
    'Salary': salaries
}

# Create a DataFrame from the dictionary
df = pd.DataFrame(data)

# Print the DataFrame
print("Manually created dataset:")
print(df)

# Save the DataFrame to a CSV file
csv_filename = 'manually_created_dataset.csv'
df.to_csv(csv_filename, index=False)
print(f"Dataset saved to '{csv_filename}' successfully.")
```

# Using Built-In Datasets in Seaborn

```python
# List of data sets in seaborn library
import seaborn as sns

# List of datasets available in seaborn
datasets_list = sns.get_dataset_names()

print("Datasets available in seaborn:")
for dataset in datasets_list:
    print(dataset)
```

```python
# Import seaborn library
import seaborn as sns

# Load the miles per gallon dataset and assign the dataset to mpg
mpg_data = sns.load_dataset('mpg')

# View the dataset (by default, the first and last five rows are displayed)
mpg_data

# print(mpg_data)
```

```python
# A sample scatter plot
# with weight on the x-axis and miles per gallon on the y-axis
# classified by origin
sns.scatterplot(data=mpg_data, x='weight', y='mpg', hue='origin')
```
